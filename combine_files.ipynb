```python
# Define initial ELO rating (common initial rating)
INITIAL_ELO = 1500

# Dictionary to store player Elo ratings for all surfaces
player_elo = {}

# Surfaces for Elo ratings
SURFACES = ['hard', 'grass', 'clay']

# Function to update Elo based on match result and surface
def update_elo(winner_elo, loser_elo, k=32):
    """
    Update Elo ratings for the winner and loser on a given surface.
    k: K-factor which determines how much the rating changes after each match.
    """
    # Calculate expected scores
    expected_winner = 1 / (1 + 10 ** ((loser_elo - winner_elo) / 400))
    expected_loser = 1 / (1 + 10 ** ((winner_elo - loser_elo) / 400))
    
    # Update ratings
    new_winner_elo = winner_elo + k * (1 - expected_winner)
    new_loser_elo = loser_elo + k * (0 - expected_loser)
    
    return new_winner_elo, new_loser_elo

# Function to get player's Elo rating for a given surface and overall
def get_player_elo(player_id, surface):
    """
    Get the current Elo rating for the player on a specific surface and overall.
    Initialize ratings if the player is new.
    """
    if player_id not in player_elo:
        # Initialize the player's ratings for all surfaces and overall
        player_elo[player_id] = {
            'total': INITIAL_ELO,
            'Hard': INITIAL_ELO,
            'Grass': INITIAL_ELO,
            'Clay': INITIAL_ELO
        }
    return player_elo[player_id]['total'], player_elo[player_id][surface]

# Process dataset row by row (assumes data is sorted by match date)
def calculate_surface_elo_for_matches(dataset):
    """
    dataset: Pandas DataFrame containing historical match data.
    Columns should include at least 'match_date', 'player1', 'player2', 'winner', 'surface'.
    """
    # List to store updated Elo ratings for each match
    elo_ratings = []

    for idx, row in dataset.iterrows():
        # Get players and surface for this match
        player1 = row['winner_id']
        player2 = row['loser_id']
        winner = row['winner_id']  # Assumes 'winner' column indicates who won the match
        surface = row['surface']  # Get the surface of the match (e.g., 'hard', 'grass', 'clay')
        
        # Get current total and surface-specific Elo ratings for both players
        player1_total_elo, player1_surface_elo = get_player_elo(player1, surface)
        player2_total_elo, player2_surface_elo = get_player_elo(player2, surface)

        # Add ELO ratings to the row for each player (total and surface-specific)
        dataset.loc[idx, 'winner_elo'] = player1_total_elo
        dataset.loc[idx, 'loser_elo'] = player2_total_elo
        dataset.loc[idx, f'winner_{surface}_elo'] = (0.5 * player1_surface_elo) + (0.5 * player1_total_elo)
        dataset.loc[idx, f'loser_{surface}_elo'] = (0.5 * player2_surface_elo) + (0.5 * player2_total_elo)
        
        # Determine the loser
        #if winner == player1:
        # Update Elo for total and specific surface ratings
        new_winner_total, new_loser_total = update_elo(player1_total_elo, player2_total_elo)
        new_winner_surface, new_loser_surface = update_elo(player1_surface_elo, player2_surface_elo)
        
        # Update Elo ratings for both players (total and surface-specific)
        player_elo[player1]['total'] = new_winner_total
        player_elo[player2]['total'] = new_loser_total
        player_elo[player1][surface] = new_winner_surface
        player_elo[player2][surface] = new_loser_surface

    return dataset
           
df_matches.sort_values('tourney_date', inplace=True)
df_matches = df_matches[df_matches['surface'] != "Carpet"]
df_matches.dropna(subset=['surface', 'w_2ndWon', 'minutes'], axis=0, inplace=True)
df_with_elo = calculate_surface_elo_for_matches(df_matches)

df_with_elo
```

```python
import math
import numpy as np

# Constants for Glicko and Glicko-2 (these can vary)
TAU = 0.5  # Glicko-2 volatility constant
INITIAL_RATING = 1500
INITIAL_RD = 350  # High RD indicates high uncertainty for new players
INITIAL_VOLATILITY = 0.06  # Initial volatility for Glicko-2

# Dictionary to store player Glicko ratings
player_glicko = {}

# Function to initialize a player's Glicko rating, RD, and volatility (for Glicko-2)
def initialize_glicko_player(player_name):
    """
    Initialize Glicko or Glicko-2 rating, RD, and volatility for a player.
    """
    if player_name not in player_glicko:
        player_glicko[player_name] = {
            'rating': INITIAL_RATING,
            'rd': INITIAL_RD,
            'volatility': INITIAL_VOLATILITY}#,  # For Glicko-2
            #'Hard': {'rating': INITIAL_RATING, 'rd': INITIAL_RD, 'volatility': INITIAL_VOLATILITY},
            #'Grass': {'rating': INITIAL_RATING, 'rd': INITIAL_RD, 'volatility': INITIAL_VOLATILITY},
            #'Clay': {'rating': INITIAL_RATING, 'rd': INITIAL_RD, 'volatility': INITIAL_VOLATILITY}}

# Function to calculate the Glicko-2 rating update
def glicko_2_update(rating, rd, volatility, outcomes, tau=TAU):
    """
    Update a player's Glicko-2 rating, RD, and volatility based on match outcomes.
    outcomes: List of tuples (opponent_rating, opponent_rd, outcome), where outcome is 1 for win, 0 for loss.
    """
    # Step-by-step implementation of Glicko-2 algorithm
    # 1. Convert ratings and RDs to Glicko-2 scale
    def g(r):
        return 1 / math.sqrt(1 + 3 * (r ** 2) / (math.pi ** 2))

    def E(player_rating, opponent_rating, opponent_rd):
        return 1 / (1 + np.exp(-g(opponent_rd) * (player_rating - opponent_rating)))

    # 2. Calculate variance (v)
    v_inv = sum([g(opponent_rd) ** 2 * E(rating, opponent_rating, opponent_rd) *
                (1 - E(rating, opponent_rating, opponent_rd)) for opponent_rating, opponent_rd, _ in outcomes])
    v = 1 / v_inv if v_inv > 0 else float('inf')

    # 3. Calculate the delta
    delta = sum([g(opponent_rd) * (outcome - E(rating, opponent_rating, opponent_rd)) for opponent_rating, opponent_rd, outcome in outcomes]) * v

    # 4. Update the volatility (for Glicko-2)
    a = math.log(volatility ** 2)
    def f(x):
        top = (np.exp(x) * (delta ** 2 - rating ** 2 - v - np.exp(x))) / (2 * (rating ** 2 + v + np.exp(x)) ** 2)
        bottom = (x - a) / (tau ** 2)
        return top - bottom

    # Using Newton's method or another technique to solve for new volatility
    # (Solving the above equation numerically)

    # 5. Update RD and rating based on new volatility
    new_rd = 1 / math.sqrt(1 / rd ** 2 + 1 / v)
    new_rating = rating + (new_rd ** 2) * sum([g(opponent_rd) * (outcome - E(rating, opponent_rating, opponent_rd)) for opponent_rating, opponent_rd, outcome in outcomes])

    return new_rating, new_rd, volatility  # Placeholder for new volatility (needs solving above)

# Process dataset row by row (assumes data is sorted by match date)
def calculate_glicko_for_matches(dataset):
    """
    dataset: Pandas DataFrame containing historical match data.
    Columns should include at least 'match_date', 'player1', 'player2', 'winner', 'surface'.
    """
    for idx, row in dataset.iterrows():
        # Get players and surface for this match
        player1 = row['winner_id']
        player2 = row['loser_id']
        #winner = row['winner_id']
        #surface = row['surface']
        
        # Initialize Glicko or Glicko-2 ratings for new players
        initialize_glicko_player(player1)
        initialize_glicko_player(player2)

        # Get current ratings, RDs, and volatilities for both players
        player1_rating, player1_rd, player1_vol = player_glicko[player1]['rating'], player_glicko[player1]['rd'], player_glicko[player1]['volatility']
        player2_rating, player2_rd, player2_vol = player_glicko[player2]['rating'], player_glicko[player2]['rd'], player_glicko[player2]['volatility']
        #print("winner init rating", player1_rating)
        #print("loser init rating", player2_rating)
        # Determine the loser
        #if winner == player1:
        outcome1, outcome2 = 1, 0

        # Collect outcomes for Glicko update
        outcomes1 = [(player2_rating, player2_rd, outcome1)]
        outcomes2 = [(player1_rating, player1_rd, outcome2)]

        # Update Glicko-2 ratings, RDs, and volatilities
        new_player1_rating, new_player1_rd, new_player1_vol = glicko_2_update(player1_rating, player1_rd, player1_vol, outcomes1)
        new_player2_rating, new_player2_rd, new_player2_vol = glicko_2_update(player2_rating, player2_rd, player2_vol, outcomes2)
        #print("new winner rating", new_player1_rating)
        #print("new loser rating", new_player2_rating)
        #if idx > 5:
        #    break
        # Update the player dictionaries with new ratings
        player_glicko[player1]['rating'], player_glicko[player1]['rd'], player_glicko[player1]['volatility'] = new_player1_rating, new_player1_rd, new_player1_vol
        player_glicko[player2]['rating'], player_glicko[player2]['rd'], player_glicko[player2]['volatility'] = new_player2_rating, new_player2_rd, new_player2_vol

        # Store the updated values in the dataset for both players
        dataset.at[idx, 'winner_glicko_rating'] = player1_rating
        dataset.at[idx, 'loser_glicko_rating'] = player2_rating
        dataset.at[idx, 'winner_glicko_rd'] = player1_rd
        dataset.at[idx, 'loser_glicko_rd'] = player2_rd
        dataset.at[idx, 'winner_glicko_volatility'] = player1_vol
        dataset.at[idx, 'loser_glicko_volatility_'] = player2_vol

    return dataset

# Assuming you have a Pandas DataFrame `df` with match history
# df = pd.read_csv('tennis_matches.csv')
df_with_elo.sort_values('tourney_date', inplace=True)

df_with_glicko = calculate_glicko_for_matches(df_with_elo)
df_with_glicko

```
